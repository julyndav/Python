## Building Cumulative Metrics

To compare performance over time, cumulative metrics are calculated for:
- Visitors
- Orders
- Revenue

This allows us to track trends and differences between test groups.

### Cumulative visitors

```python
# Adding number of visit to 'datesGroups' table
visitorsAggregated = (
    visits
    .sort_values(['group', 'date'])
    .assign(visitors=lambda x: x.groupby('group')['visits'].cumsum())
    [['date', 'group', 'visitors']]
)
visitorsAggregated
```

### Cumulative Orders

```python
# Working with the orders df
orders_daily = (
    orders
    .groupby(['date', 'group'])
    .agg(
        orders=('transactionId', 'nunique'),
        buyers=('visitorId', 'nunique'),
        revenue=('revenue', 'sum')
    )
    .reset_index()
)

ordersAggregated = (
    orders_daily
    .sort_values(['group', 'date'])
    .groupby('group')
    .apply(lambda x: x.assign(
        orders=x['orders'].cumsum(),
        buyers=x['buyers'].cumsum(),
        revenue=x['revenue'].cumsum()
    ))
    .reset_index(drop=True)
)

ordersAggregated
```

```python
# Merging the two tables into one and giving columns descriptive names
cumulativeData = ordersAggregated.merge(
    visitorsAggregated,
    on=['date', 'group']
)

cumulativeData
```

```python
# DataFrame with cumulative orders and cumulative revenue by day, group A
cumulativeRevenueA = cumulativeData[cumulativeData['group']=='A'][['date','revenue', 'orders']]

# DataFrame with cumulative orders and cumulative revenue by day, group B
cumulativeRevenueB = cumulativeData[cumulativeData['group']=='B'][['date','revenue', 'orders']]

plt.figure(figsize=(12,8))

plt.plot(cumulativeRevenueA['date'], cumulativeRevenueA['revenue'], label='Group A')

plt.plot(cumulativeRevenueB['date'], cumulativeRevenueB['revenue'], label='Group B')
plt.title('Cumulative Revenue by Group',fontdict={'size':15})
plt.xlabel('Dates',fontdict={'size':8})
plt.ylabel('Cumulative Revenue',fontdict={'size':12})
plt.grid(True)
plt.legend()
```

#### 9. Find the statistical significance of the difference in conversion between the groups using the raw data.

Analyze orders_userA and orders_usersB as data for ['userId', 'orders'], then indicate the number of orders made.

```python
# Overview of orders
orders.head()
```

```python
import scipy.stats as stats #library to work with mann whitney
from pandas.core.reshape.concat import concat # for finding the samples

# --- Step 1: Calculate number of orders per user for each group ---

# Group A: count unique orders per user
orders_userA = (
    orders[orders['group'] == 'A']
    .groupby('visitorId', as_index=False)
    .agg(orders=('transactionId', 'nunique'))
    .rename(columns={'visitorId': 'userId'})
)

# Group B: same calculation
orders_userB = (
    orders[orders['group'] == 'B']
    .groupby('visitorId', as_index=False)
    .agg(orders=('transactionId', 'nunique'))
    .rename(columns={'visitorId': 'userId'})
)


# --- Step 2: Create samples including users with zero orders ---

# Total visitors in each group
visitors_A = visits[visits['group'] == 'A']['visits'].sum()
visitors_B = visits[visits['group'] == 'B']['visits'].sum()

# Add users with zero orders to match total visitor count
sampleA = pd.concat([
    orders_userA['orders'],
    pd.Series(0, index=np.arange(visitors_A - len(orders_userA)),
              name='orders')], axis=0)
sampleB = pd.concat([
    orders_userB['orders'],
    pd.Series(0, index=np.arange(visitors_B - len(orders_userB)),
              name='orders')], axis=0)

# --- Step 3: Perform statistical test ---

# Mann-Whitney U test compares distributions between two groups
p_value = stats.mannwhitneyu(sampleA, sampleB)[1]

# Relative difference in average orders per user
relative_gain = sampleB.mean() / sampleA.mean() - 1

# --- Step 4: Output results ---

print("P-value:", f"{p_value:.4f}")
print("Relative gain (Group B vs A):", f"{relative_gain:.4f}")
```

**Visitors with zero orders were included to ensure a complete comparison.
Analyzing only users who made a purchase would bias the results and overestimate performance.**

## Group Conversion Rate Statistical Testing

Comparing the two groups (A and B) to determine whether the observed difference is real or due to random variation.

**Significance Level (α = 0.05):**  
This is the cutoff used to decide if a result is meaningful.  
It represents a 5% risk of concluding there is a difference when there is not.

**P-value:**  
The p-value measures how likely the observed results would occur if there were actually no difference between the groups.

- If p-value < 0.05 → the difference is considered statistically significant  
- If p-value ≥ 0.05 → the difference may be due to chance  

```python
#null and alternative hypotheses testing

# --- Statistical Testing ---

alpha = 0.05  # Significance level

result_grp = stats.mannwhitneyu(sampleA, sampleB)

p_value = result_grp.pvalue

print(f"P-value: {p_value:.4f}")

if p_value < alpha:
    print("Statistically significant result: the difference between groups is likely real")
else:
    print("No statistically significant difference detected")
```

**Insight:**  
The p-value (0.0168) is below the 0.05 threshold, indicating a statistically significant difference between the groups.

Group B shows a **13.8% higher conversion rate** compared to Group A, suggesting that the observed improvement is unlikely due to chance and reflects a real performance difference.

### 11. Determining the difference between groups using filtered data

```python

#-- Users with more the than (2) orders from the groups --
usersWithManyOrders = pd.concat([orders_userA[orders_userA['orders'] > 2]['userId'], #more than 2 orders for Group A
                                 orders_usersB[orders_usersB['orders'] > 2]['userId']], #more than 2 orders for Group B
                                axis = 0)
#-- With orders over $2,000 --
usersWithExpensiveOrders = orders[orders['revenue'] > 2000]['visitorId']

# Determines users that are anomalies based on order totals and revenue
abnormalUsers = pd.concat([usersWithManyOrders, usersWithExpensiveOrders],
                          axis = 0).drop_duplicates().sort_values()
print (abnormalUsers.head(5))
print("There are:",abnormalUsers.shape, "users that were found to be anomalies.")
```

The analysis showed that **22 users** came back as anomalies. They had numerous orders over $2,000. To determine how their actions affected the test results. I'll calculate the  differences in conversion between the groups, using filtered data.</b>

```python
# Orders per user for each test group.
sampleAFiltered = pd.concat([orders_userA[np.logical_not(orders_userA['userId'] \
                  .isin(abnormalUsers))]['orders'],pd.Series(0, index=np.arange(visits[visits['group']=='A']['visits'] \
                  .sum() - len(orders_userA['orders'])),name='orders')],axis=0)

sampleBFiltered = pd.concat([orders_usersB[np.logical_not(orders_usersB['userId'] \
                 .isin(abnormalUsers))]['orders'],pd.Series(0, index=np.arange(visits[visits['group']=='B']['visits'] \
                 .sum() - len(orders_usersB['orders'])),name='orders')],axis=0)
```

Applying the statistical Mann-Whitney criterion to the resulting samples.

```python
# Applying Mann-Whitney criterion to the resulting filtered samples.

print("{0:.4f}".format(stats.mannwhitneyu(sampleAFiltered, sampleBFiltered)[1]))
print("{0:.4f}".format(sampleBFiltered.mean()/sampleAFiltered.mean()-1))
```

<b>Null and Alternative Hypotheses</b>

<blockquote>alpha = 0.05 significance level
 <p>   

H0 = "There is no significant difference between the groups using filtered data."

H1 = "There is significant difference between the groups using filtered data."</blockquote>

```python
#null and alternative hypotheses testing
alpha = .05 #significance level

result_grp2 = stats.mannwhitneyu(sampleAFiltered, sampleBFiltered)

print('P-value: ','{0:.4f}'.format(result_grp2.pvalue))

if (result_grp2.pvalue < alpha):
    print("Null hypothesis rejected: the difference is statistically significant")
else:
    print("Failed to reject the null hypothesis: we can't make conclusions about the difference")
```

```python
##null and alternative hypotheses testing

alpha = .05 #significance level

#renames the PValue from above so easier to do comparison.
fos = stats.mannwhitneyu(orders[np.logical_and(
        orders['group']=='A',
        np.logical_not(orders['visitorId'].isin(abnormalUsers)))]['revenue'],
    orders[np.logical_and(
        orders['group']=='B',
        np.logical_not(orders['visitorId'].isin(abnormalUsers)))]['revenue'])[1]

print('pvalue:','{0:.4f}'.format(fos))#Prints the PValue from above.
print()

if (fos > alpha):
    print('Failed to reject null hypothesis: we can not make conclusions about the difference.')
else:
    print('Null hypothesis rejected: the difference is statisticaly significant.')
```
